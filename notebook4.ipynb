```
my_list = [3, 1, 0, 9, 5, 2, 6, 4, 9, 8, 7]
smallest = my_list[0]
for num in my_list:
    if num < smallest:
        smallest = num
print("Smallest number:", smallest)
```

```
my_list = [3, 1, 0, 9, 5, 2, 6, 4, 9, 8, 7]
total_sum = 0
count = 0
for num in my_list:
    total_sum += num
    count += 1
average = total_sum / count
print("Average:", average)
```

```
students = ["Ali", "Ahmed", "Sara", "Ayesha", "Bilal"]
user_input = input("Enter student name: ").strip()
found = False
for student in students:
    if student.lower() == user_input.lower():
        found = True
        break
if found:
    print("Student Found")
else:
    print("Student Not Found")
```

```
shopping = []
for i in range(1, 6):
    item = input(f"Enter item {i}: ").strip()
    shopping.append(item)
print("\nComplete Shopping List:")
for index, item in enumerate(shopping, 1):
    print(f"{index}. {item}")
```
